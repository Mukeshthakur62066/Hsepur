# **Data Cleaning**

**Getting all NAN Features**

```python
feature_na=[feature for feature in df.columns if df[feature].isnull().sum()>0]#it gives you all the colums that contain some null value
feature_na
```

```python
#this gives you the percentage of missing values
for feature in feature_na:
    print(feature,np.round(df[feature].isnull().mean(),4),'% missing values')
```

```python
df['rate'].unique()
```

```python
df.dropna(axis='index' ,subset=['rate'],inplace=True) #dropping the missing value
```

```python
df.shape
```

```python
def split(x):
    return x.split('/')[0]
```

```python
df['rate']=df['rate'].apply(split)
```

```python
df.head()
```

```python
df['rate'].unique()
```

```python
df.replace('NEW',0,inplace=True)
df.replace('-',0,inplace=True)
```

```python
df['rate']=df['rate'].astype(float)
```

# Finding Best budget restaurants in any location

**we will pass location and restaurant type as parameter, function will return name of restaurants.**

```python
def return_budget(loaction,restaurant):
  budget=df[(df['location']==loaction) & (df['rest_type']==restaurant) & (df['approx_cost(for two people)']<=500)]['name']
  return list(budget.unique())
```

```python
return_budget('BTM',"Quick Bites")
```
